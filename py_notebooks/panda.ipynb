## 01 - Introduction

### Data Frame
A Data Frame is a table. It contains an array of individual entries, each of which has a certain value. Each entry corresponds to a row and a column.\
`.DataFrame()`

```python
import pandas as pd

pd.DataFrame({'Bob': ['I liked it.', 'It was awful.'], 
              'Sue': ['Pretty good.', 'Bland.'],
              'Jim': ['Great!', 'So bad.']},
              index=['Product A', 'Product B'])
```

### Series
A Series, by contrast, is a sequence of data values. If a DataFrame is a table, a Series is a list. \
A Series is, in essence, a single column of a DataFrame. Assign row labels to the Series the same way as before, using an `index` parameter. \
However, a Series does not have a column name, it only has one overall `name`.  \
`.Series()`

```python
pd.Series([30, 35, 40], index=['2015 Sales', '2016 Sales', '2017 Sales'], 
          name='Product A')
```

```python
import pandas as pd
df = pd.read_csv(r"..\data\Sleep_health_and_lifestyle_dataset.csv", index_col=0)
df['BMI Category'] = df['BMI Category'].replace('Normal Weight', 'Normal')
```

### Indexing

- **Index-based Selection** \
The first is index-based selection: selecting data based on its numerical position(base 10) in the data. \
`.iloc[]` 

```python
df.iloc[0]
# selecting the first row using index [0], all columns
```

- **Label-based Selection** \
The second paradigm is the one followed by the `loc` : label-based selection. \
In this paradigm, it's the data index value(aka. the column label), not its numerical position, which matters. \
`.loc[]`

```python
df.loc[:, "Occupation"]
# referencing the Occupation column using label selection "Occupation", all rows
```

- **Manipulating the Index** \
Label-based selection derives its power from the labels in the index. Usually* the available index label is not immutable. \
We can manipulate the index labels in any way we see fit. \
`.set_index()`

```python
# df.set_index("Occupation", inplace=True)
df
```

- **is in** \
The first is `isin`. `isin` lets you select data whose value "is in" the referenced list of values.

### Maps
A map is a term for a function that takes one set of values and "maps" them to another set of values. \
We often have a need for creating new representations from existing data, or for transforming data from the format it is in now to the format that we want it to be in later.

```python
def format_sleep_duration(duration):
    temp = str(duration).split('.')
    temp[1] = str(int(float("0." + temp[1]) * 60))
    return f'{temp[0]}h {temp[1]}m'

df['Sleep Duration'].map(lambda x: format_sleep_duration(x))
```

# 04 - Grouping and Sorting


### Group By

`groupby()` splits a DataFrame into groups based on one or more columns.

Use it when you want to:
- group rows by a category
- calculate summary values for each group
- apply functions like `sum()`, `mean()`, `min()`, `max()`, or `agg()`

```python
import numpy as np
df.groupby(['Occupation'])['Quality of Sleep'].agg([len, np.mean, min, max])
# This groups the data by `Occupation` and returns basic sleep statistics for each occupation.
```

```python
df.groupby(['Quality of Sleep'])['Stress Level'].agg([len, np.mean, max, min, lambda x: max(x)-min(x)])
```
